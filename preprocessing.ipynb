# Odejścia klientów w branży telekomunikacyjnej

### Opis kolumn

- **gender** – płeć klienta 
- **SeniorCitizen** – czy klient jest seniorem (1 – tak, 0 – nie) (1, 0)
- **Partner** – czy klient jest w związku / ma partnera (Yes, No)
- **Dependents** – czy klient posiada osoby na utrzymaniu (np. dzieci) (Yes, No)
- **tenure** – staż klienta; liczba miesięcy korzystania z usług firmy
- **PhoneService** – czy klient posiada aktywną usługę telefoniczną (Yes, No)
- **MultipleLines** – czy klient posiada wiele linii telefonicznych (Yes, No, No phone service)
- **InternetService** – rodzaj dostawcy usług internetowych klienta (DSL, Fiber optic, No)
- **OnlineSecurity** – czy klient ma wykupioną usługę bezpieczeństwa online (Yes, No, No internet service)
- **OnlineBackup** – czy klient posiada usługę kopii zapasowej w chmurze (Yes, No, No internet service)
- **DeviceProtection** – czy klient posiada dodatkowe ubezpieczenie/ochronę urządzenia (Yes, No, No internet service)
- **TechSupport** – czy klient korzysta z rozszerzonego wsparcia technicznego (Yes, No, No internet service)
- **StreamingTV** – czy klient posiada usługę streamingową TV (Yes, No, No internet service)
- **StreamingMovies** – czy klient posiada usługę streamingową filmów (Yes, No, No internet service)
- **Contract** – typ zawartej umowy (Month-to-month, One year, Two year)
- **PaperlessBilling** – czy klient otrzymuje faktury w formie elektronicznej (Yes, No)
- **PaymentMethod** – preferowana metoda płatności klienta (Electronic check, Mailed check, Bank transfer (automatic), Credit card (automatic))
- **MonthlyCharges** – wysokość aktualnej miesięcznej opłaty za usługi
- **TotalCharges** – całkowita suma opłat poniesionych przez klienta od początku umowy
- **Churn** – czy klient zrezygnował z usług (Yes or No)

```python
import pandas as pd

df = pd.read_csv("data/telco_customer_churn.csv", index_col=0)
df.head()
```

```python
print(df.shape)
df.info()
```

Zamiana nazw kolumn na `snake_case`

```python
df = df.rename(columns={
    'SeniorCitizen': 'senior_citizen',
    'PhoneService': 'phone_service',
    'MultipleLines': 'multiple_lines',
    'InternetService': 'internet_service',
    'OnlineSecurity': 'online_security',
    'OnlineBackup': 'online_backup',
    'DeviceProtection': 'device_protection',
    'TechSupport': 'tech_support',
    'StreamingTV': 'streaming_tv',
    'StreamingMovies': 'streaming_movies',
    'PaperlessBilling': 'paperless_billing',
    'PaymentMethod': 'payment_method',
    'MonthlyCharges': 'monthly_charges',
    'TotalCharges': 'total_charges'
})
df.columns = df.columns.str.lower()
df.columns
```

```python
df['gender'] = df['gender'].map({'Male': 1, 'Female': 0})
df['gender']
```

```python
df['senior_citizen']
```

```python
df["partner"] = df['partner'].map({'Yes': 1, 'No': 0})
df["partner"]
```

```python
df["dependents"] = df['dependents'].map({'Yes': 1, 'No': 0})
df["dependents"]
```

```python
df["tenure"]
```

```python
df["phone_service"] = df["phone_service"].map({"Yes": 1, "No": 0})
df["phone_service"]
```

```python
cols_to_fix = [
    'multiple_lines', 'online_security', 'online_backup', 
    'device_protection', 'tech_support', 'streaming_tv', 'streaming_movies'
]

for col in cols_to_fix:
    df[col] = df[col].replace({'No internet service': 'No', 'No phone service': 'No'})
    df[col] = df[col].map({'Yes': 1, 'No': 0})

df[['multiple_lines', 'online_security', 'online_backup', 'device_protection', 'tech_support', 'streaming_tv', 'streaming_movies']]
```

```python
df = pd.get_dummies(df, columns=['internet_service'], dtype=int)
df = df.rename(columns={
    'internet_service_DSL': 'internet_service_dsl',
    'internet_service_Fiber optic': 'internet_service_fiber_optic',
    'internet_service_No': 'internet_service_no' 
})
df[['internet_service_dsl', 'internet_service_fiber_optic', 'internet_service_no']]
```

```python
df = pd.get_dummies(df, columns=['contract'], dtype=int)
df = df.rename(columns={
    'contract_Month-to-month': 'contract_month_to_month',
    'contract_One year': 'contract_one_year',
    'contract_Two year': 'contract_two_year' 
})
df[['contract_month_to_month', 'contract_one_year', 'contract_two_year']]
```

```python
df["paperless_billing"] = df["paperless_billing"].map({"Yes": 1, "No": 0})
df["paperless_billing"]
```

```python
df["payment_method"].unique()
```

```python
df = pd.get_dummies(df, columns=['payment_method'], dtype=int)
df = df.rename(columns={
    'payment_method_Bank transfer (automatic)': 'payment_method_bank_transfer_automatic',
    'payment_method_Credit card (automatic)': 'payment_method_credit_card_automatic',
    'payment_method_Electronic check': 'payment_method_electronic_check',
    'payment_method_Mailed check': 'payment_method_mailed_check'
})
df[['payment_method_bank_transfer_automatic', 'payment_method_credit_card_automatic', 
    'payment_method_electronic_check', 'payment_method_mailed_check']]
```

```python
df["monthly_charges"]
```

```python
df['total_charges'] = pd.to_numeric(df['total_charges'], errors='coerce')
df['total_charges']
```

```python
df["churn"] = df["churn"].map({"Yes": 1, "No": 0})
df["churn"]
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
temp_total_charges = pd.to_numeric(df['total_charges'], errors='coerce')
print(df[temp_total_charges.isna()][['tenure', 'total_charges']])
```

```python
df['total_charges'] = df['total_charges'].fillna(0)
df.isna().sum()
```

```python
cols = [c for c in df.columns if c != 'churn'] + ['churn']
df = df[cols]
```

```python
df.to_csv('data/preprocessed_telco_customer_churn.csv', index=True)
```

### Zaktualizowany opis kolumn

- **gender** – płeć klienta (1: Male, 0: Female).
- **senior_citizen** – czy klient ma 65+ lat (1, 0).
- **partner** – czy klient posiada partnera (1: Yes, 0: No).
- **dependents** – czy klient posiada osoby na utrzymaniu (1: Yes, 0: No).
- **tenure** – liczba miesięcy, przez które klient pozostał w firmie.
- **phone_service** – czy klient posiada usługę telefoniczną (1: Yes, 0: No).
- **multiple_lines** – czy klient posiada wiele linii (1: Yes, 0: No/No phone service).
- **online_security** – czy klient posiada zabezpieczenia online (1: Yes, 0: No/No internet service).
- **online_backup** – czy klient posiada kopię zapasową online (1: Yes, 0: No/No internet service).
- **device_protection** – czy klient posiada ochronę urządzenia (1: Yes, 0: No/No internet service).
- **tech_support** – czy klient posiada wsparcie techniczne (1: Yes, 0: No/No internet service).
- **streaming_tv** – czy klient posiada usługę streamingową TV (1: Yes, 0: No/No internet service).
- **streaming_movies** – czy klient posiada usługę streamingową filmów (1: Yes, 0: No/No internet service).
- **paperless_billing** – czy klient korzysta z fakturowania bezpapierowego (1: Yes, 0: No).
- **monthly_charges** – kwota, którą klient jest obciążany miesięcznie.
- **total_charges** – całkowita kwota, którą został obciążony klient (wartość numeryczna).
- **internet_service_dsl** – dostawca internetu to DSL (1: Yes, 0: No).
- **internet_service_fiber_optic** – dostawca internetu to światłowód (1: Yes, 0: No).
- **internet_service_no** – brak usługi internetowej (1: Yes, 0: No).
- **contract_month_to_month** – umowa miesięczna (1: Yes, 0: No).
- **contract_one_year** – umowa na rok (1: Yes, 0: No).
- **contract_two_year** – umowa na dwa lata (1: Yes, 0: No).
- **payment_method_bank_transfer_automatic** – płatność automatycznym przelewem bankowym (1: Yes, 0: No).
- **payment_method_credit_card_automatic** – płatność automatyczną kartą kredytową (1: Yes, 0: No).
- **payment_method_electronic_check** – płatność czekiem elektronicznym (1: Yes, 0: No).
- **payment_method_mailed_check** – płatność czekiem wysłanym pocztą (1: Yes, 0: No).
- **churn** – **Zmienna docelowa (target)**; czy klient odszedł (1: Yes, 0: No).
